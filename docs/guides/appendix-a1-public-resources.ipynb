# Appendix A1: Publicly Available Resources

Every ontology, SHACL shapes, and profile file [StarLayer Schemas and Ontologies](05-starlayer-schemas-and-ontologies.ipynb)'s registry bundles is also a real, individually fetchable resource - on GitHub, and as raw content - independent of whether StarLayer itself is installed on the other end. This appendix catalogs every way to reach that content directly.


## How to run this notebook

This guide assumes StarLayer has been pip installed — see [Getting Started](01-getting-started.ipynb). Run cells from top to bottom; later steps reuse variables from earlier ones.


```python
from starlayer import starontology, StarLayerGraph
```

## 1. Two different kinds of URL

`get_resource_iri(name)` is a registry entry's stable *identity* - a GitHub blob URL. Nice to browse, but a plain HTTP GET against a blob URL returns GitHub's own HTML viewer page, not the file's real bytes. `get_raw_url(name)` is a different thing entirely: a `raw.githubusercontent.com` URL that returns the entry's real content - fetchable by anything, independent of StarLayer being installed at all.

```python
print("identity (get_resource_iri):", starontology.get_resource_iri("skos_shacl"))
print("fetch   (get_raw_url):     ", starontology.get_raw_url("skos_shacl"))
```

Real content, fetched live over the network - no local file, no bundled package data involved on the reading end:

```python
remote = StarLayerGraph()
remote.parse(starontology.get_raw_url("skos_owl"), format="turtle")
print(len(remote), "triples fetched directly from GitHub")
```

## 2. The conformance-profiles mapping document

`get_raw_url()` also covers `"ontology_shapes_profiles"` - the mapping document `get_conformance_profiles()` loads (see [StarLayer Schemas and Ontologies, section 4](05-starlayer-schemas-and-ontologies.ipynb)). It has no `get_resource_iri()` counterpart at all, since it isn't one of the registry's own ontology/shapes entries:

```python
print(starontology.get_raw_url("ontology_shapes_profiles"))
```

## 3. `starlayer.shacl`'s own bundled asset files

The three SHACL asset files `starlayer.shacl` itself bundles (the meta-shapes' own presentation/validation rules, and the SHACL 1.2 conformance self-declaration from the [SHACL profiling guide](04e-shacl-profiling.ipynb)) get the identical treatment from a sibling module, `starlayer.shacl.asset_urls` - already real, standalone files, so no publish step is needed for these three, just URL-minting:

```python
from starlayer.shacl import asset_urls

for name in ("presentation_shapes", "validation_shapes", "conformance_profile"):
    print(f"{name:20} {asset_urls.get_raw_url(name)}")
```
