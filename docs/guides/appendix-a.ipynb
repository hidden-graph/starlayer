# Appendix A

An index of standalone appendix notebooks - each one a focused, self-contained reference that doesn't fit the main numbered guide sequence (1-6), either because it's a catalog rather than a walkthrough, or because it collects more advanced/complex examples that would slow down the guide they're drawn from.

- **[A1. Publicly available resources](appendix-a1-public-resources.ipynb)** - every ontology, SHACL shapes, and profile file StarLayer bundles, and how to fetch each one directly - by GitHub identity, by raw content URL, or by loading it with StarLayer installed.
- **[A2. Named, shared parts in `sqe:` query trees](appendix-a2-sqe-shared-parts.ipynb)** - giving a query part (like a repeated `FILTER`) a real, named identity instead of a private `BNode`, so more than one query in [the SPARQL queries as RDF guide](05b-sparql-query-as-rdf.ipynb)'s `sqe:` encoding can share and jointly edit it.
- **[A3. A complex DL reasoning example](appendix-a3-dl-reasoning-examples.ipynb)** - encoding a 20-person employee/job constraint-satisfaction problem as an OWL 2 DL ontology and solving it via consistency-checking alone, moved out of [the OWL 2 DL Reasoning guide](02c-owl-dl-reasoning.ipynb) to keep that guide's own examples simple.
