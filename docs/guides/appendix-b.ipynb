# Appendix B

An index of complex-example notebooks - each one a worked example too large to keep inline in the guide it's drawn from, moved out here to keep that guide's own examples simple.

- **[B1. Named, shared parts in `sqe:` query trees](appendix-b1-sqe-shared-parts.ipynb)** - giving a query part (like a repeated `FILTER`) a real, named identity instead of a private `BNode`, so more than one query in [the SPARQL queries as RDF guide](05b-sparql-query-as-rdf.ipynb)'s `sqe:` encoding can share and jointly edit it.
- **[B2. A complex DL reasoning example](appendix-b2-dl-reasoning-examples.ipynb)** - encoding a 20-person employee/job constraint-satisfaction problem as an OWL 2 DL ontology and solving it via consistency-checking alone, moved out of [the OWL 2 DL Reasoning guide](02c-owl-dl-reasoning.ipynb) to keep that guide's own examples simple.
