# Appendix A: Public Resources

StarLayer uses a number of RDF graphs internally - ontologies, SHACL shapes and a profiles document - that are also useful outside StarLayer. This appendix is a categorized catalog of those graphs and related resources, each linked to its file in the GitHub repository. 

*This appendix is a work in progress and a live document: categories and entries are added as resources are published. It contains no runnable code.*

## 2. Syntax-tree ontologies and shapes

OWL ontologies that describe the syntax tree of a language as RDF, so a parsed query, rule set or class expression can be stored, queried and edited as a graph. Each is paired with SHACL shapes that check the structural well-formedness of a graph written in that vocabulary. See [StarLayer Schemas and Ontologies](05-starlayer-schemas-and-ontologies.ipynb).

| Vocabulary | Ontology | Shapes | Description |
|---|---|---|---|
| Manchester Syntax AST (`manch:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/manchester-ast-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/manchester_shapes.ttl) | OWL 2 Manchester Syntax's own internal syntax tree as RDF - frames, clauses, class and property expressions, data ranges. See [Manchester Syntax](05a-manchester-syntax.ipynb). |
| SPARQL query edit tree (`sqe:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/sqe-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/sqe_shapes.ttl) | A SPARQL SELECT/ASK/CONSTRUCT query's raw parse tree as RDF. See [SPARQL Query as RDF](05b-sparql-query-as-rdf.ipynb). |
| SPARQL expressions and paths (`spexpr:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/spexpr-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/spexpr_shapes.ttl) | The shared expression, variable and property-path vocabulary used by the other SPARQL trees. |
| SRL (SPARQL Rules) rule sets (`srl:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/srl-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/srl_shapes.ttl) | An SRL (SPARQL Rules) rule set's own syntax tree as RDF. See [SRL (SPARQL Rules) as RDF](05c-srl-sparql-rules-as-rdf.ipynb). |

## 3. Standards formalized as RDF

The W3C's own ontology files, and StarLayer's shapes for the standards they define.

| Resource | Description |
|---|---|
| [SKOS reference ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/skos-ontology.ttl) | The W3C's own SKOS vocabulary (`skos.rdf` from the [SKOS Reference](https://www.w3.org/TR/skos-reference/), W3C Recommendation 2009), converted to Turtle with no change to its content. Used under the [W3C Software and Document License](https://www.w3.org/copyright/document-license-2023/); the copyright notice is at the top of the file. |
| [SKOS shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/skos_shapes.ttl) | StarLayer's own SHACL shapes for SKOS: the Reference's numbered conditions as validation shapes, some non-normative recommendations, and form shapes with datatype-based UI widgets. |

## 4. SHACL 1.2 meta-shapes

The SHACL-of-SHACL graph for SHACL 1.2.

| Resource | Description |
|---|---|
| [SHACL 1.2 meta-shapes (`shacl12-shacl.ttl`)](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/shacl12-shacl.ttl) | The shapes that check a shapes graph is well-formed. It combines pySHACL's bundled `shacl-shacl.ttl` (adjusted for the rules SHACL 1.2 widened) with StarLayer's supplementary rules for the SHACL vocabulary. Registry name `shacl_shacl`. |

## 5. StarLayer profiles and conformance

Defines StarLayer's own profiles and declares what each part of StarLayer conforms to.

| Resource | Description |
|---|---|
| [StarLayer profiles and conformance declaration](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/starlayer-profiles.ttl) |Each StarLayer profile (SKOS, OWL 2 Manchester Syntax, SPARQL 1.2 Query, SRL (SPARQL Rules), SHACL 1.2) is a profile of the public specification it relates to, and lists its related ontology and shapes files. Data that validates against a shapes graph conforms to the related StarLayer profile and, through that profile, to the public specification. See [SHACL profiling](04e-shacl-profiling.ipynb). Registry name `starlayer_profiles`. |
