# Appendix A: Public Resources

StarLayer uses a number of RDF graphs internally - ontologies, SHACL shapes and mapping graphs - that may also useful outside StarLayer. This appendix is a catalog of those graphs and related resources, each linked to its file in the GitHub repository.

*This appendix is a work in progress and a live document: categories and entries are added as resources are published. It contains no runnable code.*

## 1. Syntax-tree ontologies and shapes

OWL ontologies that describe the syntax tree of a language as RDF, so a parsed query, rule set or class expression can be stored, queried and edited as a graph. Each is paired with SHACL shapes that check the structural well-formedness of a graph written in that vocabulary. See [StarLayer Schemas and Ontologies](05-starlayer-schemas-and-ontologies.ipynb).

| Vocabulary | Ontology | Shapes | Description |
|---|---|---|---|
| Manchester Syntax AST (`manch:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/manchester-ast-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/manchester_shapes.ttl) | OWL 2 Manchester Syntax's syntax tree as RDF - frames, clauses, class and property expressions, data ranges. See [Manchester Syntax](05a-manchester-syntax.ipynb). |
| SPARQL query edit tree (`sqe:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/sqe-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/sqe_shapes.ttl) | A SPARQL SELECT/ASK/CONSTRUCT query's syntax tree as RDF. See [SPARQL Query as RDF](05b-sparql-query-as-rdf.ipynb). |
| SPARQL expressions and paths (`spexpr:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/spexpr-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/spexpr_shapes.ttl) | A shared expression, variable and property-path vocabulary used by the other SPARQL symtax trees. |
| SPARQL-RL rule sets (`srl:`) | [ontology](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/srl-ontology.ttl) | [shapes](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/srl_shapes.ttl) | A SRL/SPARQL-RL rule set's syntax tree as RDF. See [SPARQL Rules as RDF](05c-sparql-rules-as-rdf.ipynb). |

## 3. SHACL 1.2 meta-shapes

`shacl12-shacl.ttl` is the SHACL-of-SHACL graph for SHACL 1.2: the shapes that check a shapes graph is well-formed. It combines pySHACL's bundled `shacl-shacl.ttl` (corrected for the rules SHACL 1.2 widened) with StarLayer's supplementary rules and presentation metadata for the SHACL vocabulary, and StarLayer's validator reads it directly. See [SHACL Shapes](04-shacl-shapes.ipynb) and [SHACL UI](04d-shacl-ui.ipynb).

| Resource | Description |
|---|---|
| [SHACL 1.2 meta-shapes (`shacl12-shacl.ttl`)](https://github.com/hidden-graph/starlayer/blob/main/packages/starlayer/starlayer/starontology/shacl12-shacl.ttl) | Registry name `shacl12_shacl`. |

## 5. Fetching a resource

The links above are GitHub *blob* URLs, which return an HTML viewer page to a plain HTTP request. To fetch the actual file, use the matching `raw.githubusercontent.com` URL, or ask StarLayer for it: `starontology.get_raw_url(name)` for the entries in sections 1, 2 and 4, and `starlayer.shacl.asset_urls.get_raw_url(name)` for the SHACL conformance profile. See [StarLayer Schemas and Ontologies](05-starlayer-schemas-and-ontologies.ipynb).
