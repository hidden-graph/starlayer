## 1. Terms and Python values

Three functions in `starlayer.graph` convert between terms and Python:

- `literal(value, datatype=None, lang=None, direction=None)` makes a literal. A `bool` is `xsd:boolean`, an `int` `xsd:integer`, a `float` `xsd:double`, a `Decimal` `xsd:decimal`, and a date or time its `xsd` type.
- `to_python(term)` reads one back, and raises `ValueError` for a lexical form that is not valid for its datatype.
- `parse_term(text, prefixes=None)` reads exactly one term of Turtle 1.2 text (an IRI, a blank node, a literal, a triple term). A prefixed name needs `prefixes`.

```python
import datetime
from decimal import Decimal

from starlayer.graph import XSD, literal, parse_term, to_python

for value in (True, 5, 2.5, Decimal("1.50"), datetime.date(2026, 1, 2), "text"):
    term = literal(value)
    print(repr(value), "->", term, "->", repr(to_python(term)))

print(literal("5", XSD.long), literal("hej", lang="sv"), literal("مرحبا", lang="ar", direction="rtl"))
print(parse_term('"4"^^xsd:integer', {"xsd": str(XSD)}), parse_term("<http://example.org/a>"), parse_term("<<( <urn:s> <urn:p> <urn:o> )>>"))
```

## 7. Lists, single values and walks

These methods read and write RDF lists, keep a property single-valued, and follow a property through the graph.

- `graph.items(head)` reads an RDF list.
- `graph.add_list(items)` writes one on fresh blank nodes and returns its head (`rdf:nil` for no items).
- `graph.set_list(subject, predicate, items)` makes the items the one list a node has for a predicate, taking the cells of the old list away.
- `graph.remove_list(head)` removes the cells of a list. Its members and the triple that points at it stay.
- `graph.set(triple)` makes a triple the only one with its subject and predicate.
- `graph.transitive_objects(subject, predicate)` and `graph.transitive_subjects(predicate, object)` follow a predicate until nothing new is reached. Each node is returned once, the start node first, so a cycle ends.

```python
from starlayer.graph import RDFS, Namespace, StarLayerGraph

EXL = Namespace("http://example.org/")
lg = StarLayerGraph()
lg.set_list(EXL.order, EXL.steps, [EXL.fetch, EXL.parse, literal(3)])
print([str(t) for t in lg.items(lg.value(EXL.order, EXL.steps))])
lg.set_list(EXL.order, EXL.steps, [EXL.fetch])  # replaces the list: the old cells are gone
print(len(lg), "triples:", [str(t) for t in lg.items(lg.value(EXL.order, EXL.steps))])

lg.set((EXL.order, EXL.label, literal("first")))
lg.set((EXL.order, EXL.label, literal("second")))  # the only label now
print([str(o) for o in lg.objects(EXL.order, EXL.label)])

for sub, sup in [(EXL.Dog, EXL.Mammal), (EXL.Mammal, EXL.Animal), (EXL.Animal, EXL.Dog)]:  # a cycle
    lg.add((sub, RDFS.subClassOf, sup))
print([str(n) for n in lg.transitive_objects(EXL.Dog, RDFS.subClassOf)])
print([str(n) for n in lg.transitive_subjects(RDFS.subClassOf, EXL.Animal)])
```

## 3. Prefixes

A new graph starts with no prefixes. `StarLayerGraph(prefixes={...})` gives it some.

- `graph.bind(prefix, namespace)` adds a prefix (`override=True` replaces one that is bound), and `graph.unbind(prefix)` removes it.
- `graph.namespaces()` lists the bound prefixes, and `graph.used_prefixes()` lists the ones the graph's IRIs actually use.
- `graph.expand_curie("ex:alice")` makes the IRI of a prefixed name.
- `graph.compute_qname(iri)` splits an IRI into `(prefix, namespace, local name)`, with a `KeyError` when no prefix is bound and `generate=False`.
- `graph.qname_term(term)` writes a term as Turtle would with the graph's prefixes.

To change what every new graph starts with, call `set_default_prefixes({...})` (`None` clears it). `DEFAULT_PREFIXES` shows the current ones.

```python
import pyoxigraph as ox

from starlayer.graph import DEFAULT_PREFIXES, set_default_prefixes

pf = StarLayerGraph(prefixes={"ex": "http://example.org/"})
pf.parse(data='@prefix dc: <http://purl.org/dc/terms/> . <http://example.org/a> dc:title "T" .', format="turtle")
print("bound:", sorted(pf.namespaces()), "| used:", pf.used_prefixes())
print(pf.expand_curie("ex:alice"), pf.compute_qname(NamedNode("http://example.org/alice")), pf.qname_term(NamedNode("http://example.org/alice")))
try:
    pf.compute_qname(NamedNode("http://unknown.example/x#y"), generate=False)
except KeyError as error:
    print("KeyError:", error)
print("unbind dc:", pf.unbind("dc"), "->", sorted(pf.namespaces()))

set_default_prefixes({"ex": "http://example.org/"})
print("a new graph starts with:", sorted(StarLayerGraph().namespaces()), "| DEFAULT_PREFIXES:", dict(DEFAULT_PREFIXES))
set_default_prefixes(None)
print("after clearing:", sorted(StarLayerGraph().namespaces()))
```

## 4. Set operations

The operators `+`, `-`, `&` and `^` (union, difference, intersection and symmetric difference) work on graphs. `starlayer.graph.setops` offers the same operations as functions that also accept a bare `pyoxigraph.Store`. Blank nodes are compared by label, not by isomorphism: to ask whether two graphs are the same up to blank-node names use `isomorphic` ([6.c](06c-canonical-hashing.ipynb)).

```python
from starlayer.graph import setops

g = StarLayerGraph()
g.parse(data="@prefix ex: <http://example.org/> . ex:a ex:knows ex:b ; ex:age 4 . ex:b ex:knows ex:c .", format="turtle")
other = StarLayerGraph()
other.add((EXL.a, EXL.knows, EXL.b))
print("difference:", len(g - other), " intersection:", len(g & other), " union:", len(g + other), " symmetric:", len(g ^ other))
print("on bare stores:", len(setops.difference(g.store, other.store)))
```

## 5. Changes between graphs

`diff(old, new)` is a `GraphDelta`: the triples to add and the triples to remove to turn `old` into `new`.

- `delta.apply(graph)` makes the change, and `delta.inverse()` undoes it.
- `a.then(b)` is the net effect of `a` followed by `b`.
- `delta.applied_to(graph)` says what the triples would be without changing the graph.

Triples are compared as written, so a blank node is the same node when it has the same label. To ask whether two graphs are the same up to blank-node names use `isomorphic` ([6.c](06c-canonical-hashing.ipynb)).

```python
from starlayer.graph import diff

old, new = StarLayerGraph(prefixes={"ex": EXL}), StarLayerGraph(prefixes={"ex": EXL})
old.parse(data="@prefix ex: <http://example.org/> . ex:a ex:p ex:b ; ex:q ex:b .", format="turtle")
new.parse(data="@prefix ex: <http://example.org/> . ex:a ex:p ex:c ; ex:q ex:b .", format="turtle")

delta = diff(old, new)
show = lambda triples: sorted(" ".join(str(t) for t in triple) for triple in triples)
print("add:", show(delta.added))
print("remove:", show(delta.removed))
delta.apply(old)
print("old now equals new:", set(old.triples()) == set(new.triples()))
delta.inverse().apply(old)
print("and back:", sorted(str(o) for o in old.objects(EXL.a, EXL.p)))
```

## 6. What a graph contains

`graph.profile()` (or `profile(graph)`) reads the graph once and says what is in it:

- `classes`: each class with its number of instances.
- `properties`: each predicate with its number of triples.
- for the instances of a class, or any set of nodes, the properties they use as `PropertyUse(property, nodes, max_values, datatypes, kinds)`: how many of the nodes have the property, the most values one node has, the datatypes of its literal values, and the kinds of value (`"iri"`, `"blank"`, `"literal"`, `"triple"`).

It is a snapshot, so take another when the graph changes. A dataset is profiled by its `triples()`, which is the union of its graphs when `default_union` is on.

```python
pg = StarLayerGraph()
pg.parse(data='''@prefix ex: <http://example.org/> .
ex:a a ex:Person ; ex:name "Ann" ; ex:age 30 ; ex:knows ex:b .
ex:b a ex:Person ; ex:name "Bob", "Robert" .''', format="turtle")
profile = pg.profile()
short = lambda term: term.value.rsplit("/", 1)[-1].rsplit("#", 1)[-1]
print({short(cls): count for cls, count in profile.classes.items()})
for use in profile.class_properties(EXL.Person):
    print(short(use.property), "nodes:", use.nodes, "most values:", use.max_values, "kinds:", sorted(use.kinds), "datatypes:", sorted(short(d) for d in use.datatypes))
```

## 7. `owl:imports`

StarLayer never fetches an import on its own. You supply a `loader`, a function from an IRI to a graph (or `None`), so you decide how, and whether, a document is fetched.

- `graph.owl_imports()` lists the IRIs a graph imports. It fetches nothing.
- `graph.has_import_locally(iri)` says whether the graph already has that ontology.
- `graph.resolve_imports(loader=...)` follows the imports transitively and returns a new graph with them merged. `mode="delta"` gives only what was added, and `mode="in-place"` adds to the graph. An import that is already there, or that the loader cannot give, is skipped.
- `can_resolve_import(iri, loader=...)` asks the loader whether it would give a graph for an IRI.

`validate()` takes the same kind of loader, as `shapes_graph_loader=` ([4.e](04e-shacl-profiling.ipynb)).

```python
from starlayer.graph import can_resolve_import

BASE = NamedNode("http://example.org/base")
onto = StarLayerGraph()
onto.parse(data="<http://example.org/onto> <http://www.w3.org/2002/07/owl#imports> <http://example.org/base> .", format="turtle")

def loader(iri):  # a real loader would read a file or fetch a URL
    if iri == BASE:
        found = StarLayerGraph()
        found.parse(data="<http://example.org/base> a <http://www.w3.org/2002/07/owl#Ontology> .", format="turtle")
        return found
    return None

print("imports:", onto.owl_imports(), "| has it:", onto.has_import_locally(BASE))
print("loader can give base:", can_resolve_import(BASE, loader=loader), "| another:", can_resolve_import(NamedNode("http://example.org/none"), loader=loader))
merged = onto.resolve_imports(loader=loader)
print(len(onto), "triples before,", len(merged), "after; the merged graph has it:", merged.has_import_locally(BASE))
```

## 8. Blank nodes and copies

- `graph.skolemize()` returns a new graph in which each blank node of a subject or object (inside triple terms too) is a skolem IRI, so it can be named from outside. `authority=` and `basepath=` choose the IRI, and `bnode=` replaces only one node. `graph.de_skolemize()` turns skolem IRIs back into blank nodes.
- `graph.empty_like()` is a new empty graph of the same kind with the same prefixes (a dataset keeps its `default_union`). It is where a function that builds a result of the caller's kind starts.
- `graph.cbd(resource)` is the concise bounded description of a node: its statements and, recursively, those of its blank-node objects.

```python
from starlayer.graph import BlankNode

bn = StarLayerGraph(prefixes={"ex": "http://example.org/"})
bn.parse(data="@prefix ex: <http://example.org/> . ex:a ex:p [ ex:q 1 ] .", format="turtle")

named = bn.skolemize(authority="http://example.org", basepath="/.well-known/genid/")
print("skolem IRIs:", all(o.value.startswith("http://example.org/.well-known/genid/") for _, _, o in named.triples((None, EXL.p, None))))
restored = named.de_skolemize()
print("blank nodes again:", all(isinstance(o, BlankNode) for _, _, o in restored.triples((None, EXL.p, None))))
print("empty copy:", len(bn.empty_like()), "triples,", sorted(bn.empty_like().namespaces()))
print("cbd of ex:a:", len(bn.cbd(EXL.a)), "triples")
```
